## Diccionario de datos

PhoneService: suscripción al servicio telefónico

OnlineSecurity: suscripción adicional de seguridad en línea

OnlineBackup: suscripción adicional de respaldo en línea

DeviceProtection: suscripción adicional de protección del dispositivo

TechSupport: suscripción adicional de soporte técnico, menor tiempo de espera

StreamingMovies: suscripción de streaming de películas

Contract: tipo de contrato

PaperlessBilling: si el cliente prefiere recibir la factura en línea

PaymentMethod: forma de pago

Charges.Monthly: total de todos los servicios del cliente por mes

```python
# Crear la columna cuentas_diarias
df['cuentas_diarias'] = df['cargo_mensual'] / 30

# Ver solo las columnas relevantes
df[['cargo_mensual', 'cuentas_diarias']].head()

```

```python
# 1. Diccionario de mapeo exhaustivo para capturar todas las variantes
mapeo_completo = {
    # Binarios y variantes de "No service"
    'No': 0, 'Yes': 1,
    'No internet service': 0,
    'No phone service': 0,

    # Género
    'Female': 1, 'Male': 0,

    # Tipo de Contrato
    'Month-to-month': 0, 'One year': 1, 'Two year': 2,

    # Servicio de Internet (Convertir a categorías numéricas)
    'DSL': 1, 'Fiber optic': 2,

    # Método de Pago
    'Electronic check': 0, 'Mailed check': 1,
    'Bank transfer (automatic)': 2, 'Credit card (automatic)': 3
}

# 2. Lista de todas las columnas que actualmente son 'object' o necesitan ser numéricas
cols_a_convertir = [
    'abandono_cliente', 'genero', 'tiene_pareja', 'tiene_dependientes',
    'servicio_telefono', 'lineas_multiples', 'servicio_internet',
    'seguridad_en_linea', 'respaldo_en_linea', 'proteccion_dispositivo',
    'soporte_tecnico', 'streaming_tv', 'streaming_peliculas',
    'tipo_contrato', 'facturacion_sin_papel', 'metodo_pago'
]

# 3. Aplicar el reemplazo y forzar el tipo de dato a entero
for col in cols_a_convertir:
    if col in df.columns:
        df[col] = df[col].replace(mapeo_completo).astype(int)

# 4. Verificación final
print(df.info())
```

## Distribución por Genero

```python
# 1. Calcular frecuencias y porcentajes usando el nombre exacto de la columna
frecuencia_genero = df['genero'].value_counts()
porcentaje = df['genero'].value_counts(normalize=True) * 100

# 2. Crear el DataFrame de resumen
resumen_genero = pd.DataFrame({
    'Total': frecuencia_genero,
    'Porcentaje': porcentaje
})

# 3. Mapear los números a texto para que el reporte sea claro
# 0 -> Masculino, 1 -> Femenino
resumen_genero.index = resumen_genero.index.map({0: 'Masculino', 1: 'Femenino'})

# 4. Aplicar los formatos de miles y porcentaje
resumen_genero['Total'] = resumen_genero['Total'].apply(lambda x: f'{x:,.0f}')
resumen_genero['Porcentaje'] = resumen_genero['Porcentaje'].apply(lambda x: f'{x:,.2f} %')

# Mostrar el resultado
resumen_genero
```

```python
# 1. Preparar los datos
resumen_grafico = resumen_genero.copy()
resumen_grafico['Porcentaje_Num'] = resumen_grafico['Porcentaje'].str.replace(' %', '').astype(float)

# 2. Configuración estética
plt.figure(figsize=(8, 5))
colores = ['#3d647e', '#4da078'] # Azul y Verde de tu identidad visual

# 3. Crear el gráfico de barras
ax = sns.barplot(
    x=resumen_grafico.index,
    y=resumen_grafico['Porcentaje_Num'],
    palette=colores,
    hue=resumen_grafico.index,
    legend=False
)

# 4. Añadir etiquetas de porcentaje sobre las barras
for p in ax.patches:
    ax.annotate(f'{p.get_height():.2f}%',
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha = 'center', va = 'center',
                xytext = (0, 9),
                textcoords = 'offset points',
                fontsize=11, fontweight='bold')

# 5. Personalización final
plt.title('Distribución de Clientes por Género', fontsize=14, pad=20, fontweight='bold')
plt.ylabel('Porcentaje (%)', fontsize=12)
plt.xlabel('Género', fontsize=12)
plt.ylim(0, 100) # El tope es 100%
sns.despine() # Quita los bordes innecesarios

plt.show()
```

```python
# 1. Crear los rangos de antigüedad (bins)
bins = [0, 6, 12, 24, 36, 48, 72]
labels = ['0-6 meses', '7-12 meses', '13-24 meses', '25-36 meses', '37-48 meses', '49-72 meses']
df['rango_antiguedad'] = pd.cut(df['antiguedad_meses'], bins=bins, labels=labels)

# 2. Calcular la tasa de deserción por rango
tasa_desercion = df.groupby('rango_antiguedad', observed=False)['abandono_cliente'].mean() * 100

# 3. Crear el gráfico de línea
plt.figure(figsize=(12, 5))
plt.plot(tasa_desercion.index, tasa_desercion.values, marker='o', color='gray', linewidth=2)

# 4. Estilo y etiquetas
plt.title('Tasa de deserción por antigüedad del cliente', loc='left', fontsize=16, fontweight='bold')
plt.suptitle('Los clientes con una antigüedad menor a 6 meses presentan la mayor tasa de deserción',
             x=0.45, y=0.88, fontsize=12, color='gray')

# Resaltar el primer punto en rojo y añadir etiquetas de porcentaje
for i, valor in enumerate(tasa_desercion.values):
    color_punto = 'red' if i == 0 else 'gray'
    plt.scatter(tasa_desercion.index[i], valor, color=color_punto, zorder=5)
    plt.text(tasa_desercion.index[i], valor + 2, f'{valor:.1f}%',
             ha='center', fontweight='bold', color=color_punto)

# Limpiar estética del gráfico
plt.ylim(0, max(tasa_desercion.values) + 10)
sns.despine(left=True)
plt.yticks([])
plt.xlabel('Antigüedad')
plt.show()
```

Curva de Estabilización: Existe una relación inversa clara; a mayor antigüedad, menor es el riesgo de fuga. Al llegar al rango de 49-72 meses, la tasa cae drásticamente al 9.5%, lo que sugiere que los esfuerzos de retención deben enfocarse agresivamente en los primeros 12 meses.

## 4. Conclusiones e Insights del Abandono

La tasa global de abandono es del **26.6%**, y el análisis revela los siguientes puntos críticos:

- **Riesgo por Precio:** Los clientes que abandonan presentan una mediana de pago de $79.7, significativamente superior a los $64.5 de los clientes retenidos. El precio elevado es el principal detonante de fuga.  
- **Tipo de Contrato:** El contrato *Mes a Mes* muestra el mayor riesgo, con una tasa de abandono del 42%. En contraste, los contratos a 1 y 2 años funcionan como un mecanismo efectivo de retención.  
- **Vulnerabilidad Etaria:** La tasa de permanencia en adultos mayores desciende al 58.32%, frente al 76.35% observado en otros grupos etarios.  
- **Métodos de Pago:** El 45.3% de las bajas se concentra en clientes que utilizan transferencia bancaria. Los métodos digitales presentan una mayor velocidad de abandono.  
- **Servicios Críticos:** El servicio de internet de alta velocidad es la principal fuente de ingresos, pero también el más vulnerable frente a la competencia.

## 5. El Punto Crítico de Supervivencia

El análisis temporal muestra que la etapa más vulnerable corresponde a los **primeros 6 meses** de relación con el cliente. Durante este periodo ocurre la mayor proporción de abandonos.

**Hallazgo Clave:** El riesgo de deserción aumenta significativamente cuando coinciden cargos mensuales elevados (mayores a $70) con una baja antigüedad del cliente.

## 6. Recomendaciones Estratégicas

Con base en los hallazgos, se proponen las siguientes acciones:

- **Programa de Bienvenida (0–6 meses):** Implementar incentivos de lealtad durante el primer semestre para superar el punto crítico de deserción.  
- **Migración de Contratos:** Diseñar campañas comerciales que incentiven el cambio de contratos *Mes a Mes* hacia contratos anuales mediante descuentos estratégicos.  
- **Ajuste de Valor en Planes Premium:** Revisar la propuesta de valor para clientes con cargos superiores a $70, mejorando la percepción de beneficio frente al precio.  
- **Atención Especializada:** Crear un canal de soporte diferenciado para adultos mayores y usuarios de internet de alta velocidad.  
- **Optimización de Pagos:** Analizar y reducir fricciones asociadas al método de transferencia bancaria para disminuir el abandono en este segmento.
